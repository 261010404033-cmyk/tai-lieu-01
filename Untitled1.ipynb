```python
# %% [Ô 3] Chạy toàn bộ quy trình & Xuất file minh chứng cho báo cáo
import cv2
import torch
import easyocr
from ultralytics import YOLO
from huggingface_hub import hf_hub_download

CONF_THRESHOLD = 0.2
PADDING = 5

# --- Đọc ảnh gốc ---
image = cv2.imread(IMAGE_PATH)
if image is None:
    raise FileNotFoundError(f"Không đọc được ảnh: {IMAGE_PATH}")
h_img, w_img = image.shape[:2]
annotated_image = image.copy()  # Tạo bản sao để vẽ khung minh họa

# --- Bước 1: Tải và chạy mô hình YOLO DocLayNet ---
model_path = hf_hub_download(repo_id="hantian/yolo-doclaynet", filename="yolov8n-doclaynet.pt")
model = YOLO(model_path)
results = model.predict(source=image, conf=CONF_THRESHOLD, verbose=False)

boxes = []
for box in results[0].boxes:
    x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())
    boxes.append((x1, y1, x2, y2))

# --- Bước 2: Sắp xếp tăng dần theo y1 (từ trên xuống dưới) ---
boxes.sort(key=lambda b: b[1])

# --- Bước 3: Vẽ khung đánh số & Cắt ảnh ---
crops = []
for idx, (x1, y1, x2, y2) in enumerate(boxes, start=1):
    # Vẽ khung xanh và số thứ tự đọc lên ảnh minh họa
    cv2.rectangle(annotated_image, (x1, y1), (x2, y2), (0, 200, 0), 2)
    cv2.putText(annotated_image, f"#{idx}", (x1, max(y1 - 8, 15)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

    # Cắt ảnh với lề an toàn (PADDING)
    cx1, cy1 = max(x1 - PADDING, 0), max(y1 - PADDING, 0)
    cx2, cy2 = min(x2 + PADDING, w_img), min(y2 + PADDING, h_img)
    crop = image[cy1:cy2, cx1:cx2]
    if crop.size > 0:
        crops.append(crop)

if not crops:
    crops = [image]

# Lưu ảnh minh họa ra file
cv2.imwrite("anh_nhan_dien.jpg", annotated_image)

# --- Bước 4: EasyOCR đọc chữ và lưu file TXT ---
reader = easyocr.Reader(["vi"], gpu=torch.cuda.is_available())
full_text = []
for crop in crops:
    lines = reader.readtext(crop, detail=0, paragraph=True)
    text = "\n".join(lines).strip()
    if text:
        full_text.append(text)

final_output = "\n\n".join(full_text)

with open("ket_qua_van_ban.txt", "w", encoding="utf-8") as f:
    f.write(final_output)

print("✅ Hoàn tất! Đã lưu 'anh_nhan_dien.jpg' và 'ket_qua_van_ban.txt' ở cột thư mục bên trái.")
```
